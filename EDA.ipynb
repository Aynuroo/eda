```python
df = pd.read_csv('/content/Global YouTube Statistics.csv', encoding='latin1')    #UnicodeDecodeError   erroru verirdi deye encoding ederek fayli oxuyuruq
df
```

```python
#bu sutunlarin vizuallarda gostermeyin lazimsiz oldugunu dusunduyum ucun silirem

df.drop(['Title', 'Abbreviation','video_views_rank','country_rank', 'channel_type_rank', 'video_views_for_the_last_30_days','subscribers_for_last_30_days', 'Gross tertiary education enrollment (%)', 'Population', 'Unemployment rate', 'Urban_population', 'Latitude', 'Longitude'], axis = 1, inplace = True)
```

```python
df
```

```python
copy = df.copy()  #sehvlik oldugu zaman geri bu df e qayitmaq ucun kopyasini yaratdim
```

```python
df["created_year"] = df["created_year"].astype("Int64")    #data type'larini deyisirik, cunki integer olmalidir
df["created_date"] = df["created_date"].astype("Int64")
```

```python
df['created_year'][(df['created_year'] <= 2005) | (df['created_year'] >= 2026)] = None   #uygunsuz olan deyerleri null edirik
df['created_date'][(df['created_date'] <= 1) | (df['created_date'] >= 31)] = None
```

```python
df.isna().sum()
```

```python
df["created_month"] = pd.to_datetime(                      #numeric edirik
    df["created_month"], format="%b", errors="coerce"
).dt.month
```

```python
df["created_year"] = df["created_year"].fillna(df["created_year"].median())                     #null deyerleri silmirik, az oldugu ucun doldura bilerik, median ve mode ile doldurmusam
df["created_month"] = df["created_month"].fillna(df["created_month"].mode()[0])                 #niye median? -> cunki, outlier'lara davamlidir, mean ile etmek duzgun netice almamagimiza sebeb ola biler
df["created_date"] = df["created_date"].fillna(df["created_date"].median())                     #niye mode? -> cunki, en cox tekrarlanan deyer ile evezlemek en ideal secimdir burada
```

```python
df.isna().sum()
```

```python
df["created_datetime"] = pd.to_datetime(
    df.rename(columns={
        "created_year": "year",
        "created_month": "month",                       #bu 3 sutunda verilmis il, ay, gun olanlari birlesdirib 1 sutunda yigdim
        "created_date": "day"
    })[["year", "month", "day"]],
    errors="coerce"
)
```

```python
df = df.drop(columns=["created_year", "created_month", "created_date"])  #daha sonra bu column'lari silirik
```

```python
df.duplicated().sum()   #dublicate data yoxdur ona gore deymirem buna
```

```python
df["category"] = df["category"].fillna("Unknown")
df["Country"] = df["Country"].fillna("Unknown")                #null deyerleri unknown ile doldururam, cunki median ve ya mode ile yalandan yazmaq vizual gostericilere yanlis tesir gostermeyine sebeb ola biler
df["channel_type"] = df["channel_type"].fillna("Unknown")
```

```python
df.isna().sum()
```

```python
df["estimated_yearly_earnings"] = (
    df["lowest_yearly_earnings"] + df["highest_yearly_earnings"]       #illik midpoint tapiram vizualda gostermek ucun
) / 2
```

```python
df["estimated_monthly_earnings"] = (
    df["lowest_monthly_earnings"] + df["highest_monthly_earnings"]     #eyni sekilde ayliq ucun de tapiram
) / 2
```

```python
df = df.drop(columns=["highest_yearly_earnings", "lowest_yearly_earnings", "highest_monthly_earnings", "lowest_monthly_earnings"])  #daha sonra bu column'lari silirik
```

```python
df
```

```python
import matplotlib.pyplot as plt
                                                                      #Hansı Kateqoriya Daha Çox Qazanır? (illik)
category_revenue = (                                                  #vizualdan da gorunduyu kimi en cox qazandiran kateqoriya entertainment olub,
    df[df["category"] != "Unknown"]                                   # en az ise travel & events
    .groupby("category")["estimated_yearly_earnings"]
    .sum()
    .sort_values()

)

category_revenue.plot(kind="barh", figsize=(9,5), color="red")
plt.title("Total Estimated Yearly Earnings by Category")
plt.xlabel("Estimated Yearly Earnings")
plt.ylabel("Category")
plt.show()
```

```python
plt.scatter(df["estimated_monthly_earnings"], df["rank"], alpha=0.6, color="red")            #ayliq qiymət vs Rating
plt.title("Monthly earnings vs Rating")                                                      #daha yuksek gelir, daha yaxsi rank demekdir
plt.xlabel("Monthly earnings")
plt.ylabel("Rating")
plt.show()
```

```python
df.set_index("created_datetime")["estimated_monthly_earnings"] \
    .resample("M") \
    .mean() \
    .plot(color="red", figsize=(10,5))

plt.title("Zaman üzrə Orta Aylıq Gəlir Trendi")
plt.ylabel("Orta Aylıq Gəlir")
plt.xlabel("Tarix")
plt.show()

#en yuksek qazanc 2006ci ilde olub, 2009-2017 araliginda stabillik olub,
#daha sonra yeniden yukselik olub
```

```python
plt.scatter(df["subscribers"], df["video views"], alpha=0.5, color='red')          #Views vs Subscribers
plt.xlabel("Subscribers")
plt.ylabel("Video Views")
plt.title("Subscribers vs Views")
plt.show()

#abone sayi artiqca, baxis sayi artir, ve gorunduyu kimi outlier deyerler de movcuddur
```

```python
top_countries = (
    df[df["Country"] != "Unknown"]
    .groupby("Country")["subscribers"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_countries.plot(kind="bar", color='red', figsize=(10,5))
plt.title("Top 10 Countries by Subscribers (Known Only)")
plt.ylabel("Subscribers")
plt.xlabel("Country")
plt.xticks(rotation=45, ha='right')
plt.show()


#Ölkələr üzrə TOP 10
#USA en cox aboneye sahib olan olkedir
```

```python
# Top 5 abunəçi olan kanallar
top_5_subs = df[['Youtuber', 'subscribers']].sort_values(
    by='subscribers', ascending=False
).head(5)

# Bar chart
plt.figure(figsize=(8,5))
plt.bar(top_5_subs['Youtuber'], top_5_subs['subscribers'], color='red')
plt.title("Top 5 Youtuber by Subscribers")
plt.ylabel("Subscribers")
plt.xlabel("Youtuber")
plt.xticks(rotation=45, ha='right')
plt.show()

# Top 5 abunəçiyə malik youtuber
top_5_subs = df[['Youtuber', 'subscribers']].sort_values(
    by='subscribers', ascending=False
).head(5)

top_5_subs
```
